# 2. Weekly graph question

Below are the graphs from last week's weekly graph question.  Suppose you wanted to cause the viewer to examine the last three months of data first.  How could you use preattentive attributes to do that?  Suggest several possibilities.

**My answer**

Preattentive attributes are the visual properties the brain processes almost instantly (in well under a second), *before* conscious attention kicks in — color, intensity, size, line weight, enclosure, position, orientation, and added markers. To pull the viewer's eye to the last three months first, I'd lean on one or two of these rather than all of them (if everything is emphasized, nothing is). Several possibilities:

1. **Color / hue** — the single strongest cue. Draw the first nine months in a muted grey and the last three months in one bold, saturated accent color. The eye jumps straight to the colored segment.

2. **Intensity / saturation** — same idea even within one color: keep the early line pale and make only the recent stretch dark and high-contrast.

3. **Line weight (size)** — thicken the last-three-months segment so it reads as heavier and more important than the thinner earlier line.

4. **Enclosure** — drop a light shaded band (a soft rectangle) behind the final three months. Enclosure groups those points and signals "look here" without changing the line itself.

5. **Added markers** — put dots (or larger dots) only on the last three data points, so they pop out against an unmarked earlier line.

6. **Position / a divider** — add a thin vertical reference line at month 9 with a small "last quarter" label, using whitespace and a boundary to set the recent window apart.

7. **Direct labeling / text** — annotate the last point with its value or a short callout. A bit of text set apart from the rest is itself preattentive and anchors the eye.

In practice I'd combine just **color + one supporting cue** — e.g., a grey line with the final three months in accent color *and* a light enclosure band — because a single dominant attribute does the work and stacking too many competing signals dilutes the effect.

```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt

# Storytelling With Data, ch.4 ("focus your audience's attention"): the signature
# preattentive-attribute move - a mostly-grey line with one segment in bold color,
# a light enclosure band, and a text callout pulling the eye to the recent months.
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
x = list(range(12))
revenue = [42, 45, 44, 48, 47, 51, 50, 53, 52, 58, 64, 71]  # ticks up at the end

fig, ax = plt.subplots(figsize=(10, 5.5))

# Soft enclosure band behind the last three months (preattentive: enclosure)
ax.axvspan(8.5, 11.5, color="#1f6fb2", alpha=0.08, zorder=0)

# Early months muted grey; final three months bold accent + thicker (color + size)
ax.plot(x[:9],  revenue[:9],  color="#bdbdbd", linewidth=2.5, zorder=2)
ax.plot(x[8:],  revenue[8:],  color="#1f6fb2", linewidth=3.5, zorder=3)
ax.scatter(x[9:], revenue[9:], color="#1f6fb2", s=45, zorder=4)  # markers on recent points

# Text callout (preattentive: text set apart)
ax.annotate("Strong finish:\n+22% in the last quarter",
            xy=(11, revenue[-1]), xytext=(8.6, 40),
            color="#1f6fb2", fontsize=11, fontweight="bold")

# Declutter
ax.set_xticks(x); ax.set_xticklabels(months)
for s in ["top", "right"]:
    ax.spines[s].set_visible(False)
for s in ["left", "bottom"]:
    ax.spines[s].set_color("#cccccc")
ax.tick_params(colors="#666666")
ax.set_ylim(35, 78)
ax.set_ylabel("Revenue ($K)", color="#666666")
ax.set_title("Look at the last three months first", loc="left", fontsize=14, color="#333333")
plt.tight_layout(); plt.show()
```
