# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

**My answer**

The first version draws the two weeks as **two side-by-side panels**; the second **overlays both weeks on a single set of axes**.

**Side-by-side panels — advantages:** each week is uncluttered and easy to read on its own, and the shape of each week's pattern is clear. **Disadvantages:** comparing the two is hard — the eye has to jump between panels, and unless the y-axes are locked to the same range the comparison is actively misleading (a difference in scale looks like a difference in visitors). The gap between panels also breaks the "day of week" x-axis into two disconnected pieces.

**Overlaid on one axis — advantages:** this is the better choice here. Both weeks share the same axes, so you can directly see where week 2 was higher or lower than week 1 day-by-day — the comparison is immediate. It uses less space and keeps a single, continuous day-of-week axis. **Disadvantages:** it needs a legend or direct labels to tell the two lines apart, and if you tried to overlay many weeks (not just two) the lines would turn into unreadable spaghetti.

**Which I'd use:** for *comparing* two weeks, the overlaid single-axis version — with the two lines labeled directly and drawn in clearly different colors. I'd only prefer separate panels if the two weeks were on genuinely different scales or if the goal were to present each week in isolation rather than compare them.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import numpy as np

# Storytelling With Data, ch.8 ("pulling it all together"): a single polished bar chart
# that uses everything - decluttered frame, one accent color, direct value labels,
# a takeaway title + supporting subtitle, and a source note.
depts  = ["Support", "Sales", "Product", "Finance", "Legal", "HR"]
scores = [88, 84, 79, 71, 64, 61]
accent = "#1f6fb2"; grey = "#cfcfcf"
colors = [accent if s >= 80 else grey for s in scores]

fig, ax = plt.subplots(figsize=(9, 5.5))
bars = ax.barh(depts[::-1], scores[::-1], color=colors[::-1], height=0.66)
for b, s in zip(bars, scores[::-1]):
    ax.text(s + 1, b.get_y()+b.get_height()/2, f"{s}", va="center", color="#555555", fontsize=11)

for sp in ["top", "right", "bottom"]:
    ax.spines[sp].set_visible(False)
ax.spines["left"].set_color("#cccccc")
ax.tick_params(left=False, bottom=False, labelbottom=False, colors="#333333")
ax.grid(False); ax.set_xlim(0, 100)

# Title hierarchy: big takeaway, smaller supporting line
ax.text(0, 6.8, "Support and Sales lead on employee engagement",
        fontsize=15, fontweight="bold", color="#333333", transform=ax.transData)
ax.text(0, 6.4, "2024 engagement survey score by department (0–100)",
        fontsize=11, color="#888888", transform=ax.transData)
ax.text(0, -1.1, "Source: Annual employee engagement survey, n = 1,240",
        fontsize=9, color="#aaaaaa", transform=ax.transData)

ax.set_ylim(-1.4, 7.2)
plt.tight_layout(); plt.show()
```
